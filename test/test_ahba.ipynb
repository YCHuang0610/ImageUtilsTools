```python
expression = process.get_expression
sample_info = process.get_sample_info
```

```python
import pandas as pd

donors = [9861, 10021, 12876, 14380, 15496, 15697]
donor_expression = {}
for donor in donors:
    donor_expression[donor] = expression[sample_info['brain'] == donor].groupby(
        sample_info['atlas_label']
    ).mean()

agg_expression = pd.concat(donor_expression)
agg_expression
```

```python
df = agg_expression.groupby('atlas_label').mean()
```
